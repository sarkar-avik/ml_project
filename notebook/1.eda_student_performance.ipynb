### 3.6 Check various categories present in the different categorical column

```python
categorical_columns = [feature for feature in df.columns if df[feature].dtype == 'str']
numerical_columns = [feature for feature in df.columns if df[feature].dtype != 'str']
```

```python
print("There are {} numerical features: {}".format(len(numerical_columns), numerical_columns))
print("There are {} numerical features: {}".format(len(categorical_columns), categorical_columns))
```

```python
for i in range(len(categorical_columns)):
    print(f"Categories in {categorical_columns[i]}: {df[categorical_columns[i]].unique().tolist()}")
```

### 3.7 Adding columns for Total Score and Average

```python
df['total_score'] = df['math_score'] + df['reading_score'] + df['writing_score']
df['average'] = df['total_score']/len(numerical_columns)
df.head()
```

```python
math_full = df[df['math_score'] == 100]['average'].count()
writing_full = df[df['writing_score'] == 100]['average'].count()
reading_full = df[df['reading_score'] == 100]['average'].count()

print("Number of students with full marks in Math:      ", math_full)
print("Number of students with full marks in Writing:   ", writing_full)
print("Number of students with full marks in Reading:   ", reading_full)
```

```python
math_less_20 = df[df['math_score'] <= 20]['average'].count()
writing_less_20 = df[df['writing_score'] <= 20]['average'].count()
reading_less_20 = df[df['reading_score'] <= 20]['average'].count()

print("Number of students with full marks in Math:      ", math_less_20)
print("Number of students with full marks in Writing:   ", writing_less_20)
print("Number of students with full marks in Reading:   ", reading_less_20)
```

### Insights

- From above values we get students have performed the worst in Maths
- Best performance is in reading section
